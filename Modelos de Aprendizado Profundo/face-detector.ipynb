```python
caminho_cascade = '/home/alan-rocha/anaconda3/pkgs/opencv-4.12.0-py313h204cae0_0/share/opencv4/haarcascades/haarcascade_frontalface_default.xml'
```

```python
import cv2

# Carrega o classificador Haar de rostos que já vem junto com o OpenCV
detector_rosto = cv2.CascadeClassifier(caminho_cascade)

# Confirma que carregou (deve imprimir True)
print("Classificador carregado?", not detector_rosto.empty())

# Abre a webcam. O índice 0 é a câmera padrão
camera = cv2.VideoCapture(0)

# Abre a webcam. O índice 0 é a câmera padrão
if not camera.isOpened():
    print("Não consegui acessar a webcam. Verifique se ela está conectada/liberada.")
else:
    print('Webcam ligada!! Aperte "q" na janela de vídeo para sair.')

    while True:
        # lê um quadro (frame) da câmera
        ok, quadro = camera.read()
        if not ok:
            break

        # O detector trabalha em tons de cinza, então convertemos:
        cinza = cv2.cvtColor(quadro, cv2.COLOR_BGR2GRAY)

        # Detecta os rostos. Retorna uma lista de retângulos (x, y, largura, altura)
        rostos = detector_rosto.detectMultiScale(
            cinza,
            scaleFactor = 1.1,
            minNeighbors = 5,
            minSize = (60, 60)
        )

        # Desenha um retângulo em volta de cada rosto encontrado
        for (x, y, w, h) in rostos:
            cv2.rectangle(quadro, (x, y), (x + w, y + h), (0, 255, 0), 2)

        # Mostra quantos rostos foram detectados no canto da tela
        cv2.putText(quadro, f'Rostos: {len(rostos)}', (10, 30),
                    cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

        # Exibe o resultado na janela
        cv2.imshow('Deteccao de Rostos - aperte q para sair', quadro)

        # Sai do loop quando 'q' é pressionado
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    # Libera a câmera e fecha as janelas
    camera.release()
    cv2.destroyAllWindows()
    
```
